```python
import warnings
warnings.filterwarnings('ignore')
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.utils import to_categorical
import numpy as np

print("TensorFlow version:", tf.__version__)

# Load MNIST data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
print(f"Training samples: {len(x_train)}, Test samples: {len(x_test)}")

# Preprocess: resize to 96x96 (smaller, faster) and convert to RGB
print("Preprocessing data...")
x_train = x_train.reshape(-1, 28, 28, 1).astype('float32') / 255.0
x_train = np.repeat(x_train, 3, axis=-1)  # Grayscale to RGB
x_train = tf.image.resize(x_train, (96, 96)).numpy()

x_test = x_test.reshape(-1, 28, 28, 1).astype('float32') / 255.0
x_test = np.repeat(x_test, 3, axis=-1)
x_test = tf.image.resize(x_test, (96, 96)).numpy()

# One-hot encode labels
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print("Building model...")
# Load pre-trained MobileNetV2 (lighter than EfficientNet)
base_model = MobileNetV2(
    weights='imagenet', 
    include_top=False, 
    input_shape=(96, 96, 3)
)

# Unfreeze last 30 layers
base_model.trainable = True
for layer in base_model.layers[:-30]:
    layer.trainable = False

print(f"Total layers: {len(base_model.layers)}")
print(f"Trainable layers: {sum([1 for layer in base_model.layers if layer.trainable])}")

# Build model
model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

# Compile
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("\nStarting training...")
# Train with smaller batch size to save memory
history = model.fit(
    x_train, y_train,
    validation_data=(x_test, y_test),
    epochs=10,
    batch_size=64,  # Increased from 32 for faster training
    verbose=1
)

# Evaluate
print("\nEvaluating model...")
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f'\n{"="*50}')
print(f'Final Test Accuracy: {accuracy*100:.2f}%')
print(f'Expected: 97-98%')
print(f'{"="*50}')
```
